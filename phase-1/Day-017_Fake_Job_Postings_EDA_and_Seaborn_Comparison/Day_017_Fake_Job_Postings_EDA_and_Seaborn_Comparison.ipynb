```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats
```

```python
df = pd.read_csv('/content/fake_job_postings.csv')
df.head()
```

```python
df.tail()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df["required_experience"] = df["required_experience"].fillna("Unknown")
df["employment_type"] = df["employment_type"].fillna("Unknown")
df["required_education"] = df["required_education"].fillna("Unknown")
```

```python
df.hist(figsize=(20, 15))
plt.show()
```

```python
sns.countplot(x='required_experience', hue='fraudulent', data=df)
plt.title('Fraudulent Jobs by Experience')
plt.xticks(rotation=45)
plt.show()
```

```python
sns.countplot(x='employment_type', hue='fraudulent', data=df)
plt.title('Fraudulent Jobs by Employment Type')
plt.xticks(rotation=45)
plt.show()
```

```python
sns.countplot(x='required_education', hue='fraudulent', data=df)
plt.title('Fraudulent Jobs by Education')
plt.xticks(rotation=45)
plt.show()
```

```python
sns.countplot(x='telecommuting', hue='fraudulent', data=df)
plt.title('Fraudulent Jobs by Remote Work')
plt.show()
```
