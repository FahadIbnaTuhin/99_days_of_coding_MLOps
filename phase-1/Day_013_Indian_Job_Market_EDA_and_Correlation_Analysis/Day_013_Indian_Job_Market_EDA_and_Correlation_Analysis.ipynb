```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats
```

```python
df = pd.read_excel('/content/indian-job-market-dataset-2025.xlsx')
df.head()
```

```python
df.tail()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df["minimumExperience"] = df["minimumExperience"].fillna(df["minimumExperience"].mean())
df["minimumSalary"] = df["minimumSalary"].fillna(df["minimumSalary"].mean())
df["maximumSalary"] = df["maximumSalary"].fillna(df["maximumSalary"].mean())
df["AggregateRating"] = df["AggregateRating"].fillna(df["AggregateRating"].mean())
df["ReviewsCount"] = df["ReviewsCount"].fillna(df["ReviewsCount"].mean())
df["maximumExperience"] = df["maximumExperience"].fillna(df["maximumExperience"].mean())
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
df.hist(figsize=(20,15))
plt.show()
```

```python
sns.scatterplot(x='minimumExperience', y='minimumSalary', data=df)
plt.title("minimumExperience vs minimumSalary")
plt.show()
```

```python
sns.scatterplot(x='minimumExperience', y='maximumSalary', data=df)
plt.title("minimumExperience vs maximumSalary")
plt.show()
```

```python
sns.scatterplot(x='maximumExperience', y='minimumSalary', data=df)
plt.title("maximumExperience vs minimumSalary")
plt.show()
```

```python
sns.scatterplot(x='maximumExperience', y='maximumSalary', data=df)
plt.title("maximumExperience vs maximumSalary")
plt.show()
```

```python
# X-axis: minimumExperience → How many years of experience the job requires.
# Y-axis: AggregateRating → How highly employees rate that company (e.g. 3.5 out of 5).

sns.scatterplot(x='minimumExperience', y='AggregateRating', data=df)
plt.title("minimumExperience vs AggregateRating")
plt.show()
```

```python
sns.scatterplot(x='maximumExperience', y='AggregateRating', data=df)
plt.title("maximumExperience vs AggregateRating")
plt.show()
```

```python
# X-axis: minimumExperience → Required years of experience.
# Y-axis: ReviewsCount → How many employee reviews the company has.

sns.scatterplot(x='minimumExperience', y='ReviewsCount', data=df)
plt.title("minimumExperience vs ReviewsCount")
plt.show()
```

```python
sns.scatterplot(x='maximumExperience', y='ReviewsCount', data=df)
plt.title("maximumExperience vs ReviewsCount")
plt.show()
```
